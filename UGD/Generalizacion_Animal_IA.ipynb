```python
import numpy as np
from sklearn.neural_network import MLPClassifier

# Caracteristicas: [tamano_orejas, largo_cola, tiene_bigotes, sonido_frecuencia]
# Etiqueta (ultima columna): 0=Perro, 1=Gato
animales_entrenamiento = [
    [0.3, 0.4, 1, 0.8, 1],  # Gato domestico
    [0.2, 0.5, 1, 0.9, 1],  # Gato domestico 2
    [0.4, 0.3, 1, 0.7, 1],  # Gato domestico 3
    [0.6, 0.8, 0, 0.3, 0],  # Perro mediano
    [0.7, 0.9, 0, 0.2, 0],  # Perro grande
]

nombres_ejemplos = [
    "Gato domestico 1", "Gato domestico 2", "Gato domestico 3",
    "Perro mediano", "Perro grande",
]

nombres_clases = ["Perro", "Gato"]

datos = np.array(animales_entrenamiento)
X_train = datos[:, :4]
y_train = datos[:, 4].astype(int)

print("ENTRENAMIENTO")
print("=" * 50)
for nombre, fila, etiqueta in zip(nombres_ejemplos, X_train, y_train):
    print(f"  {nombre}: {fila} -> {nombres_clases[etiqueta]}")

modelo = MLPClassifier(hidden_layer_sizes=(5,), max_iter=1000, random_state=42)
modelo.fit(X_train, y_train)
print(f"\nPrecision en entrenamiento: {modelo.score(X_train, y_train):.2%}")

casos_nuevos = np.array([
    [0.25, 0.45, 1, 0.85],  # Gato persa
    [0.35, 0.4, 1, 0.75],   # Gato siames
    [0.8, 1.0, 0, 0.1],     # Pastor aleman
    [0.5, 0.2, 1, 0.6],     # Gato manx (cola corta)
])

descripciones = [
    "Gato PERSA (raza nueva)",
    "Gato SIAMES (raza nueva)",
    "Pastor ALEMAN (raza nueva)",
    "Gato MANX (cola corta, atipico)",
]

print("\nGENERALIZACION A CASOS NUEVOS")
print("=" * 50)
predicciones = modelo.predict(casos_nuevos)
probabilidades = modelo.predict_proba(casos_nuevos)

for desc, fila, pred, prob in zip(descripciones, casos_nuevos, predicciones, probabilidades):
    print(f"  {desc}: {fila}")
    print(f"    -> Prediccion: {nombres_clases[pred]}  "
          f"(P(Perro)={prob[0]:.2%}, P(Gato)={prob[1]:.2%})")

print("\nIMPORTANCIA APROXIMADA DE CADA CARACTERISTICA")
print("=" * 50)
nombres_features = ["tamano_orejas", "largo_cola", "tiene_bigotes", "sonido_frecuencia"]
peso_promedio = np.mean(np.abs(modelo.coefs_[0]), axis=1)
for nombre, peso in zip(nombres_features, peso_promedio):
    print(f"  {nombre}: {peso:.3f}")
```
